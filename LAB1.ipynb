```python
print("AR TRAVEL & TOURS")

name = input("Enter your name: ")
destination = input("Enter your destination: ")
budget = float(input("Enter your budget (PKR): "))

print("\nChoose your travel preference:")
print("1. Budget")
print("2. Comfortable")
print("3. Luxury")

preference = input("Enter your choice (1-3): ")

# Agent decision-making
if preference == "1":
     hotel = "Budget Hotel"
     transport = "Public Transport"

elif preference == "2":
     hotel = "3-Star Hotel"
     transport = "Private/Shared Transport"

elif preference == "3":
     hotel = "5-Star Hotel"
     transport = "Private Car"

else:
     print("Invalid preference. Defaulting to Budget plan.")
     hotel = "Budget Hotel"
     transport = "Public Transport"

# Budget evaluation
if budget < 50000:
   budget_status = "Low budget"
elif budget <= 150000:
   budget_status = "Medium budget"
else:
   budget_status = "High budget"

# Final recommendation
print("\nAR TRAVEL & TOURS")
print(f"\nTraveler: {name}")
print(f"Destination: {destination}")
print(f"Budget: PKR {budget:,.0f}")
print(f"Budget Category: {budget_status}")
print(f"Hotel: {hotel}")
print(f"Transport: {transport}")

print("\nTravel Agent: Plan generated successfully!")
```
